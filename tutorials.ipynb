```python
import matplotlib.pyplot as plt
from matplotlib_venn import venn2, venn3

# Unique alpha_beta values per method
sets = {
    method: set(group["alpha_beta"].dropna().unique())
    for method, group in result.groupby("method")
}

methods = list(sets)

if len(methods) == 2:
    venn2(
        [sets[methods[0]], sets[methods[1]]],
        set_labels=methods
    )
elif len(methods) == 3:
    venn3(
        [sets[methods[0]], sets[methods[1]], sets[methods[2]]],
        set_labels=methods
    )
else:
    raise ValueError("Venn diagrams are best for 2–3 methods.")

plt.title("Shared unique alpha_beta values between methods")
plt.show()
```
